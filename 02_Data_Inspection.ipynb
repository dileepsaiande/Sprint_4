# Notebook 2 – Data Inspection

## Objective
The objective of this notebook is to inspect the Titanic dataset and understand its structure, dimensions, column names, data types, index, and basic statistics.

## Topics Covered
- Number of rows and columns
- Data types and column names
- Index, shape, head(), tail(), and sample()
- info() and describe()
- dtypes and nunique()
- value_counts()
- Identification of numerical, categorical, date/time, identifier, and potential target columns

```python
import pandas as pd

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
display(df.head())
```

```python
print("Number of rows:", len(df))
print("Number of columns:", len(df.columns))
print("Dataset shape:", df.shape)
```

```python
print("Column names:")
for column in df.columns:
    print(column)
```

```python
print("Dataset index:")
print(df.index)

print("\nFirst 10 index values:")
print(df.index[:10])
```

```python
print("First 5 rows:")
display(df.head())

print("Last 5 rows:")
display(df.tail())

print("5 random rows:")
display(df.sample(5, random_state=42))
```

```python
print("Dataset information:")
df.info()
```

```python
print("Data type of each column:")
display(df.dtypes)
```

```python
print("Numerical column statistics:")
display(df.describe())

print("Summary of all columns:")
display(df.describe(include="all"))
```

```python
print("Unique values in each column:")
display(df.nunique())
```

```python
print("Passenger class counts:")
display(df["Pclass"].value_counts())

print("Survival counts:")
display(df["Survived"].value_counts())

print("Sex counts:")
display(df["Sex"].value_counts())

print("Embarkation port counts:")
display(df["Embarked"].value_counts(dropna=False))
```

```python
import pandas as pd

df = pd.DataFrame({
    "PassengerId": [1, 2, 3, 4, 5],
    "Age": [22, 38, 26, 35, 28],
    "Sex": ["male", "female", "female", "female", "male"],
    "Survived": [0, 1, 1, 1, 0]
})

print("Rows and columns:", df.shape)
display(df)
```

```python
print("Column names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)
```

```python
print("First 5 rows:")
display(df.head())

print("Last 5 rows:")
display(df.tail())

print("Dataset information:")
df.info()

print("Statistical summary:")
display(df.describe(include="all"))
```

```python
print("Unique values in each column:")
display(df.nunique())
```

```python
print("Passenger ID counts:")
display(df["PassengerId"].value_counts())

print("\nSex counts:")
display(df["Sex"].value_counts())

print("\nSurvival counts:")
display(df["Survived"].value_counts())
```

```python
print("FINAL DATA INSPECTION SUMMARY")
print("-----------------------------")

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isnull().sum().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nData inspection completed successfully!")
```
