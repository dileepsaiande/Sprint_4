```python
import seaborn as sns
import pandas as pd

df = sns.load_dataset("tips")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

```python
# Identify numerical variables
numerical_columns = df.select_dtypes(include="number").columns.tolist()

# Identify categorical variables
categorical_columns = df.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("Numerical Variables:")
print(numerical_columns)

print("\nCategorical Variables:")
print(categorical_columns)

print("\nTotal numerical variables:", len(numerical_columns))
print("Total categorical variables:", len(categorical_columns))
```

```python
df.describe()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Select the first numerical column
column = numerical_columns[0]

plt.figure(figsize=(8, 5))
sns.histplot(data=df, x=column, bins=20, kde=True)

plt.title(f"Distribution of {column}")
plt.xlabel(column)
plt.ylabel("Frequency")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

column = numerical_columns[0]

plt.figure(figsize=(8, 4))
sns.boxplot(data=df, x=column)

plt.title(f"Box Plot of {column}")
plt.xlabel(column)
plt.show()
```

```python
# Calculate descriptive statistics for numerical columns

for column in numerical_columns:
    print(f"\nStatistics for: {column}")
    print("-" * 40)

    print("Mean:", df[column].mean())
    print("Median:", df[column].median())
    print("Mode:", df[column].mode().tolist())
    print("Variance:", df[column].var())
    print("Standard Deviation:", df[column].std())
    print("Minimum:", df[column].min())
    print("Maximum:", df[column].max())
    print("Range:", df[column].max() - df[column].min())
```

```python
# Calculate skewness and kurtosis for numerical variables

for column in numerical_columns:
    print(f"\nAnalysis for: {column}")
    print("-" * 40)

    print("Skewness:", df[column].skew())
    print("Kurtosis:", df[column].kurt())
```

```python
# Analyze categorical variables

for column in categorical_columns:
    print(f"\nAnalysis for: {column}")
    print("-" * 40)

    print("Number of unique values:", df[column].nunique())
    print("\nFrequency counts:")
    print(df[column].value_counts(dropna=False))
```

```python
# Calculate category percentages

for column in categorical_columns:
    print(f"\nPercentages for: {column}")
    print("-" * 40)

    percentages = (
        df[column]
        .value_counts(normalize=True, dropna=False)
        .mul(100)
        .round(2)
    )

    print(percentages)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

for column in categorical_columns:
    plt.figure(figsize=(8, 5))

    sns.countplot(data=df, x=column, order=df[column].value_counts().index)

    plt.title(f"Count Plot of {column}")
    plt.xlabel(column)
    plt.ylabel("Count")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
for column in categorical_columns:
    percentages = (
        df[column].value_counts(normalize=True, dropna=False) * 100
    )

    plt.figure(figsize=(8, 5))
    percentages.plot(kind="bar")

    plt.title(f"Percentage Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Percentage (%)")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
import matplotlib.pyplot as plt

for column in categorical_columns:
    counts = df[column].value_counts().head(8)

    if len(counts) > 0:
        plt.figure(figsize=(7, 7))

        plt.pie(
            counts.values,
            labels=counts.index.astype(str),
            autopct="%1.1f%%",
            startangle=90
        )

        plt.title(f"Pie Chart of {column}")
        plt.axis("equal")
        plt.show()
```

```python
print("UNIVARIATE ANALYSIS SUMMARY")
print("=" * 40)

print("Total rows:", df.shape[0])
print("Total columns:", df.shape[1])
print("Numerical variables:", len(numerical_columns))
print("Categorical variables:", len(categorical_columns))

print("\nNumerical columns:")
print(numerical_columns)

print("\nCategorical columns:")
print(categorical_columns)
```

```python
# Count missing values in each column
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)

print("\nTotal missing values:", df.isnull().sum().sum())
```

```python
# Count duplicate rows
print("Number of duplicate rows:", df.duplicated().sum())
```
