```python
import pandas as pd
import seaborn as sns

try:
    df = sns.load_dataset("titanic")
    print("Dataset loaded successfully!")
    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])
    display(df.head())
except Exception as e:
    print("Dataset loading failed:", e)
```

```python
print("Dataset information:")
df.info()

print("\nMissing values:")
display(df.isnull().sum().to_frame("Missing Values"))

print("\nDuplicate rows:", df.duplicated().sum())
```

```python
# 1. Check dataset dimensions
print("Dataset shape:", df.shape)

# 2. Check column names
print("\nColumn names:")
print(df.columns.tolist())

# 3. Check data types and non-null counts
print("\nDataset information:")
df.info()

# 4. Count missing values
print("\nMissing values in each column:")
display(df.isnull().sum().to_frame("Missing Values"))

# 5. Check duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())

# 6. Check basic statistics
print("\nNumerical summary:")
display(df.describe())

print("\nCategorical summary:")
display(df.describe(include="object"))
```

```python
# Step 1: Count missing values in each column
missing_values = df.isnull().sum()

print("Missing values before cleaning:")
display(missing_values[missing_values > 0].sort_values(ascending=False))

# Step 2: Calculate missing-value percentages
missing_percent = (df.isnull().sum() / len(df)) * 100

print("\nMissing-value percentages:")
display(
    missing_percent[missing_percent > 0]
    .sort_values(ascending=False)
    .round(2)
)

# Step 3: Create a cleaned copy
df_clean = df.copy()

# Step 4: Fill missing numerical values with the median
numeric_cols = df_clean.select_dtypes(include="number").columns

for col in numeric_cols:
    if df_clean[col].isnull().any():
        df_clean[col] = df_clean[col].fillna(df_clean[col].median())

# Step 5: Fill missing categorical values with the mode
categorical_cols = df_clean.select_dtypes(
    include=["object", "str", "category"]
).columns

for col in categorical_cols:
    if df_clean[col].isnull().any():
        mode_values = df_clean[col].mode()
        if not mode_values.empty:
            df_clean[col] = df_clean[col].fillna(mode_values.iloc[0])

# Step 6: Verify remaining missing values
print("\nRemaining missing values:")
display(df_clean.isnull().sum()[df_clean.isnull().sum() > 0])

print("\nOriginal dataset shape:", df.shape)
print("Cleaned dataset shape:", df_clean.shape)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Select numerical columns
numeric_cols = df_clean.select_dtypes(include="number").columns

# Calculate outliers using the IQR method
outlier_summary = []

for col in numeric_cols:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    count = (
        (df_clean[col] < lower_limit) |
        (df_clean[col] > upper_limit)
    ).sum()

    outlier_summary.append({
        "Column": col,
        "Lower Limit": lower_limit,
        "Upper Limit": upper_limit,
        "Outlier Count": count
    })

outlier_df = pd.DataFrame(outlier_summary)

print("Outlier analysis:")
display(outlier_df.sort_values("Outlier Count", ascending=False))
```

```python
# Display boxplots for numerical columns
for col in numeric_cols:
    plt.figure(figsize=(8, 3))
    sns.boxplot(x=df_clean[col])
    plt.title(f"Outlier Detection: {col}")
    plt.xlabel(col)
    plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Count survivors and non-survivors
print("Survival counts:")
display(df_clean["survived"].value_counts().rename(
    index={0: "Did not survive", 1: "Survived"}
))

# 2. Calculate survival percentages
print("\nSurvival percentages:")
display(
    (df_clean["survived"].value_counts(normalize=True) * 100).round(2)
)

# 3. Visualize target distribution
plt.figure(figsize=(6, 4))
sns.countplot(data=df_clean, x="survived")
plt.title("Target Variable Distribution")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")
plt.show()

# 4. Compare survival with passenger class
print("\nSurvival rate by passenger class:")
display(
    df_clean.groupby("class", observed=True)["survived"]
    .mean()
    .mul(100)
    .round(2)
    .rename("Survival Rate (%)")
)

# 5. Compare survival by sex
print("\nSurvival rate by sex:")
display(
    df_clean.groupby("sex", observed=True)["survived"]
    .mean()
    .mul(100)
    .round(2)
    .rename("Survival Rate (%)")
)
```

### Target Variable Analysis — Observations

1. The target variable is `survived`.
2. A value of 0 represents a passenger who did not survive; 1 represents a passenger who survived.
3. Survival counts and percentages help assess target class balance.
4. Grouping survival by passenger class and sex helps identify differences between passenger groups.
5. These are descriptive relationships and do not, by themselves, establish causation.

```python
# Create a new copy for feature engineering
df_featured = df_clean.copy()

# 1. Create age groups
df_featured["age_group"] = pd.cut(
    df_featured["age"],
    bins=[0, 12, 18, 35, 60, float("inf")],
    labels=["Child", "Teenager", "Young Adult", "Adult", "Senior"],
    include_lowest=True
)

# 2. Create family size
df_featured["family_size"] = (
    df_featured["sibsp"] + df_featured["parch"] + 1
)

# 3. Identify passengers traveling alone
df_featured["traveling_alone"] = (
    df_featured["family_size"] == 1
)

# 4. Create fare categories
df_featured["fare_category"] = pd.cut(
    df_featured["fare"],
    bins=[-1, 10, 30, 100, float("inf")],
    labels=["Low", "Medium", "High", "Very High"]
)

# Display the new features
print("New features created successfully!")

display(
    df_featured[
        ["age", "age_group", "sibsp", "parch",
         "family_size", "traveling_alone", "fare", "fare_category"]
    ].head(10)
)
```

```python
# Survival rate by age group
print("Survival rate by age group:")
display(
    df_featured.groupby("age_group", observed=True)["survived"]
    .mean()
    .mul(100)
    .round(2)
    .rename("Survival Rate (%)")
)

# Survival rate by family size
print("\nSurvival rate by family size:")
display(
    df_featured.groupby("family_size")["survived"]
    .mean()
    .mul(100)
    .round(2)
    .rename("Survival Rate (%)")
)

# Visualize survival by family size
plt.figure(figsize=(8, 4))
sns.barplot(
    data=df_featured,
    x="family_size",
    y="survived"
)
plt.title("Survival Rate by Family Size")
plt.ylabel("Average Survival Rate")
plt.show()
```

```python
import pandas as pd

# Use the feature-engineered dataset
df_model = df_featured.copy()

# Remove rows where the target is missing
df_model = df_model.dropna(subset=["survived"])

# Separate target and input features
X = df_model.drop(columns=["survived"])
y = df_model["survived"]

# One-hot encode categorical columns
X_encoded = pd.get_dummies(
    X,
    columns=X.select_dtypes(
        include=["object", "str", "category", "bool"]
    ).columns,
    drop_first=True,
    dtype=int
)

print("Original feature shape:", X.shape)
print("Encoded feature shape:", X_encoded.shape)

print("\nEncoded feature columns:")
print(X_encoded.columns.tolist())

display(X_encoded.head())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)

# Build a basic classification pipeline
model = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("classifier", LogisticRegression(max_iter=1000))
])

# Train the model
model.fit(X_train, y_train)

# Make predictions
y_pred = model.predict(X_test)

# Evaluate the model
print("\nAccuracy:", round(accuracy_score(y_test, y_pred), 4))

print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import (
    ConfusionMatrixDisplay,
    RocCurveDisplay,
    roc_auc_score
)

# 1. Plot confusion matrix
ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred,
    display_labels=["Did Not Survive", "Survived"],
    cmap="Blues"
)

plt.title("Confusion Matrix")
plt.show()

# 2. Plot ROC curve
y_probability = model.predict_proba(X_test)[:, 1]

RocCurveDisplay.from_predictions(
    y_test,
    y_probability
)

plt.title("ROC Curve")
plt.show()

# 3. Calculate ROC-AUC score
auc_score = roc_auc_score(y_test, y_probability)

print("ROC-AUC Score:", round(auc_score, 4))
```
